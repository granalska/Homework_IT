```python
from collections import deque

def add_edge(graph, capacity, first, second, value):
    #додаємо зв'язок між вершинами
    if second not in graph[first]:
        graph[first].append(second)

    if first not in graph[second]:
        graph[second].append(first)

    capacity[first, second] = value
    capacity[second, first] = 0


def edmonds_karp(graph, capacity, start, finish):
    #створюємо потік
    flow = {}

    for edge in capacity:
        flow[edge] = 0

    maximum_flow = 0

    #шукаємо шляхи поки вони є
    while True:
        parent = {}
        parent[start] = None

        queue = deque()
        queue.append(start)

        while len(queue) > 0:

            current = queue.popleft()

            for next_vertex in graph[current]:

                if next_vertex not in parent:

                    current_capacity = capacity.get(
                        (current, next_vertex), 0)

                    current_flow = flow.get(
                        (current, next_vertex), 0)

                    if current_capacity - current_flow > 0:
                        parent[next_vertex] = current
                        queue.append(next_vertex)

        #якщо кінцева вершина не знайдена
        if finish not in parent:
            break

        #знаходимо найменшу пропускну здатність
        path_flow = 1000000

        current = finish

        while current != start:
            previous = parent[current]

            available = capacity.get(
                (previous, current), 0) - flow.get(
                (previous, current), 0)

            if available < path_flow:
                path_flow = available

            current = previous

        #змінюємо потік
        current = finish

        while current != start:

            previous = parent[current]

            if (previous, current) not in flow:
                flow[previous, current] = 0

            if (current, previous) not in flow:
                flow[current, previous] = 0

            flow[previous, current] += path_flow
            flow[current, previous] -= path_flow

            current = previous

        maximum_flow += path_flow

    return maximum_flow, flow


#створюємо граф
graph = {}

vertices = [
    'Source',
    'Terminal 1',
    'Terminal 2',
    'Warehouse 1',
    'Warehouse 2',
    'Warehouse 3',
    'Warehouse 4']

for i in range(1, 15):
    vertices.append('Shop ' + str(i))

vertices.append('Sink')


#створюємо порожні списки
for vertex in vertices:
    graph[vertex] = []


capacity = {}


#термінал 1
add_edge(graph, capacity, 'Terminal 1', 'Warehouse 1', 25)
add_edge(graph, capacity, 'Terminal 1', 'Warehouse 2', 20)
add_edge(graph, capacity, 'Terminal 1', 'Warehouse 3', 15)


#термінал 2
add_edge(graph, capacity, 'Terminal 2', 'Warehouse 3', 15)
add_edge(graph, capacity, 'Terminal 2', 'Warehouse 4', 30)
add_edge(graph, capacity, 'Terminal 2', 'Warehouse 2', 10)


#склад 1
add_edge(graph, capacity, 'Warehouse 1', 'Shop 1', 15)
add_edge(graph, capacity, 'Warehouse 1', 'Shop 2', 10)
add_edge(graph, capacity, 'Warehouse 1', 'Shop 3', 20)


#склад 2
add_edge(graph, capacity, 'Warehouse 2', 'Shop 4', 15)
add_edge(graph, capacity, 'Warehouse 2', 'Shop 5', 10)
add_edge(graph, capacity, 'Warehouse 2', 'Shop 6', 25)


#склад 3
add_edge(graph, capacity, 'Warehouse 3', 'Shop 7', 20)
add_edge(graph, capacity, 'Warehouse 3', 'Shop 8', 15)
add_edge(graph, capacity, 'Warehouse 3', 'Shop 9', 10)


#склад 4
add_edge(graph, capacity, 'Warehouse 4', 'Shop 10', 20)
add_edge(graph, capacity, 'Warehouse 4', 'Shop 11', 10)
add_edge(graph, capacity, 'Warehouse 4', 'Shop 12', 15)
add_edge(graph, capacity, 'Warehouse 4', 'Shop 13', 5)
add_edge(graph, capacity, 'Warehouse 4', 'Shop 14', 10)


#додаємо початок і кінець
add_edge(graph, capacity, 'Source', 'Terminal 1', 60)
add_edge(graph, capacity, 'Source', 'Terminal 2', 55)

for i in range(1, 15):
    add_edge(
        graph,
        capacity,
        'Shop ' + str(i),
        'Sink',
        100)


#запускаємо алгоритм
maximum_flow, flow = edmonds_karp(
    graph,
    capacity,
    'Source',
    'Sink')


print('Максимальний потік:', maximum_flow)
print('--------------------------\n')


#виводимо потоки між терміналами і магазинами
print('Термінал - Магазин - Фактичний потік')
print('--------------------------')


for terminal in ['Terminal 1', 'Terminal 2']:

    for i in range(1, 15):

        shop = 'Shop ' + str(i)

        #знаходимо через який склад йде потік
        shop_flow = 0

        if terminal == 'Terminal 1':

            if i == 1:
                shop_flow = flow.get(
                    ('Warehouse 1', 'Shop 1'), 0)

            if i == 2:
                shop_flow = flow.get(
                    ('Warehouse 1', 'Shop 2'), 0)

            if i == 3:
                shop_flow = flow.get(
                    ('Warehouse 1', 'Shop 3'), 0)

            if i == 4:
                shop_flow = flow.get(
                    ('Warehouse 2', 'Shop 4'), 0)

            if i == 5:
                shop_flow = flow.get(
                    ('Warehouse 2', 'Shop 5'), 0)

            if i == 6:
                shop_flow = flow.get(
                    ('Warehouse 2', 'Shop 6'), 0)

            if i == 7:
                shop_flow = flow.get(
                    ('Warehouse 3', 'Shop 7'), 0)

            if i == 8:
                shop_flow = flow.get(
                    ('Warehouse 3', 'Shop 8'), 0)

            if i == 9:
                shop_flow = flow.get(
                    ('Warehouse 3', 'Shop 9'), 0)

        print(
            terminal,
            '-',
            shop,
            '-',
            shop_flow)
print('--------------------------\n')
```

```python

#створюємо клас Trie
class Trie:
    class Node:
        def __init__(self):
            self.children = {}
            self.is_end = False
    def __init__(self):
        self.root = self.Node()
    def put(self, word, value):
        current = self.root
        for letter in word:
            if letter not in current.children:
                current.children[letter] = self.Node()
            current = current.children[letter]
        current.is_end = True
class Homework(Trie):
    def count_words_with_suffix(self, pattern) -> int:

        #перевіряємо введені дані
        if not isinstance(pattern, str):
            raise TypeError('Потрібно ввести рядок')
        if len(pattern) == 0:
            raise ValueError('Рядок не може бути пустим')
        words = []

        #функція для пошуку слів
        def find_words(node, word):
            if node.is_end:
                words.append(word)
            for letter in node.children:
                next_node = node.children[letter]
                new_word = word + letter
                find_words(next_node, new_word)
        find_words(self.root, '')
        count = 0

        #перевіряємо всі слова
        for word in words:
            if word.endswith(pattern):
                count = count + 1
        return count
    def has_prefix(self, prefix) -> bool:

        #перевіряємо дані
        if not isinstance(prefix, str):
            raise TypeError('Потрібно ввести рядок')
        if len(prefix) == 0:
            raise ValueError('Префікс не може бути пустим')
        current = self.root

        #перевіряємо букви префікса
        for letter in prefix:
            if letter not in current.children:
                return False
            current = current.children[letter]
        return True
if __name__ == '__main__':

    #створюємо дерево
    trie = Homework()

    #додаємо слова
    words = [
        'apple',
        'application',
        'banana',
        'cat']
    for i, word in enumerate(words):
        trie.put(word, i)


    #перевіряємо суфікси
    print('Слів на e:',
          trie.count_words_with_suffix('e'))
    print('Слів на ion:',
          trie.count_words_with_suffix('ion'))
    print('Слів на a:',
          trie.count_words_with_suffix('a'))
    print('Слів на at:',
          trie.count_words_with_suffix('at'))
    print('--------------------------\n')


    #перевіряємо префікси
    print('Префікс app:',
          trie.has_prefix('app'))
    print('Префікс bat:',
          trie.has_prefix('bat'))
    print('Префікс ban:',
          trie.has_prefix('ban'))
    print('Префікс ca:',
          trie.has_prefix('ca'))
    print('--------------------------\n')

    #перевіряємо правильність результатів
    assert trie.count_words_with_suffix('e') == 1
    assert trie.count_words_with_suffix('ion') == 1
    assert trie.count_words_with_suffix('a') == 1
    assert trie.count_words_with_suffix('at') == 1
    assert trie.has_prefix('app') == True
    assert trie.has_prefix('bat') == False
    assert trie.has_prefix('ban') == True
    assert trie.has_prefix('ca') == True
    print('Перевірка пройдена')
```
