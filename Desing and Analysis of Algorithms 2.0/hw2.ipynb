```python
from typing import List, Tuple

def find_min_max(numbers: List[float]) -> Tuple[float, float]:
    #перевірка масиву
    if len(numbers) == 0:
        raise ValueError('Масив не може бути пустим')

    #якщо один елемент
    if len(numbers) == 1:
        return numbers[0], numbers[0]

    #якщо два елементи
    if len(numbers) == 2:
        if numbers[0] < numbers[1]:
            return numbers[0], numbers[1]
        else:
            return numbers[1], numbers[0]

    #ділимо масив на дві частини
    middle = len(numbers) // 2

    left_numbers = numbers[:middle]
    right_numbers = numbers[middle:]

    #знаходимо мінімум і максимум для лівої частини
    left_minimum, left_maximum = find_min_max(left_numbers)

    #знаходимо мінімум і максимум для правої частини
    right_minimum, right_maximum = find_min_max(right_numbers)

    #знаходимо загальний мінімум
    if left_minimum < right_minimum:
        minimum = left_minimum
    else:
        minimum = right_minimum

    #знаходимо загальний максимум
    if left_maximum > right_maximum:
        maximum = left_maximum
    else:
        maximum = right_maximum

    return minimum, maximum


#тест масиву
numbers = [15, 7, 23, 4, 18, 9, 31, 2, 12]

minimum, maximum = find_min_max(numbers)

print('Масив:', numbers)
print('----------------------------\n')
print('Мінімальний елемент:', minimum)
print('----------------------------\n')
print('Максимальний елемент:', maximum)
print('----------------------------\n')
```
