```python
from typing import List, Tuple


def find_min_max(numbers: List[float]) -> Tuple[float, float]:
    #перевіряємо чи масив не пустий
    if len(numbers) == 0:
        raise ValueError('Масив не може бути пустим')

    #якщо тільки один елемент
    if len(numbers) == 1:
        minimum = numbers[0]
        maximum = numbers[0]

        return minimum, maximum

    #ділимо масив на дві частини
    middle = len(numbers) // 2

    left_numbers = numbers[:middle]
    right_numbers = numbers[middle:]

    #знаходимо мінімум і максимум першої частини
    left_minimum, left_maximum = find_min_max(left_numbers)

    #знаходимо мінімум і максимум другої частини
    right_minimum, right_maximum = find_min_max(right_numbers)

    #порівнюємо мінімальні значення
    if left_minimum < right_minimum:
        minimum = left_minimum
    else:
        minimum = right_minimum

    #порівнюємо максимальні значення
    if left_maximum > right_maximum:
        maximum = left_maximum
    else:
        maximum = right_maximum

    return minimum, maximum


#тест масиву
numbers = [15, 7, 23, 4, 18, 9, 31, 2, 12]

minimum, maximum = find_min_max(numbers)

print('Масив:', numbers)
print('Мінімальний елемент:', minimum)
print('Максимальний елемент:', maximum)
print('--------------------------\n')
```

```python
from typing import List, Dict
from dataclasses import dataclass


@dataclass
class PrintJob:
    id: str
    volume: float
    priority: int
    print_time: int


@dataclass
class PrinterConstraints:
    max_volume: float
    max_items: int


def optimize_printing(print_jobs: List[Dict], constraints: Dict) -> Dict:
    #отримуємо обмеження принтера
    max_volume = constraints['max_volume']
    max_items = constraints['max_items']

    #створюємо новий список
    jobs = []

    for job in print_jobs:
        jobs.append(job)

    #сортуємо завдання за пріоритетом
    for i in range(len(jobs)):
        for j in range(len(jobs) - 1):
            if jobs[j]['priority'] > jobs[j + 1]['priority']:
                temporary_job = jobs[j]
                jobs[j] = jobs[j + 1]
                jobs[j + 1] = temporary_job

    print_order = []
    total_time = 0

    current_jobs = []
    current_volume = 0
    current_time = 0

    #перебираємо всі завдання
    for job in jobs:

        job_id = job['id']
        job_volume = job['volume']
        job_time = job['print_time']

        can_add = True

        #перевіряємо кількість моделей
        if len(current_jobs) >= max_items:
            can_add = False

        #перевіряємо об'єм
        if current_volume + job_volume > max_volume:
            can_add = False

        #якщо модель можна додати
        if can_add:

            current_jobs.append(job)
            current_volume = current_volume + job_volume

            #визначаємо час групи
            if job_time > current_time:
                current_time = job_time

        else:

            #якщо група вже має моделі
            if len(current_jobs) > 0:

                #додаємо моделі до порядку друку
                for current_job in current_jobs:
                    print_order.append(current_job['id'])

                #додаємо час групи
                total_time = total_time + current_time

            #створюємо нову групу
            current_jobs = []
            current_volume = 0
            current_time = 0

            #додаємо поточну модель в нову групу
            current_jobs.append(job)
            current_volume = job_volume
            current_time = job_time

    #додаємо останню групу
    if len(current_jobs) > 0:

        for current_job in current_jobs:
            print_order.append(current_job['id'])

        total_time = total_time + current_time

    #створюємо результат
    result = {
        'print_order': print_order,
        'total_time': total_time}

    return result


#тест моделі однакового пріоритету
test1_jobs = [
    {'id': 'M1', 'volume': 100, 'priority': 1, 'print_time': 150},
    {'id': 'M2', 'volume': 150, 'priority': 1, 'print_time': 100},
    {'id': 'M3', 'volume': 120, 'priority': 1, 'print_time': 150}]


#тест моделі різних пріоритетів
test2_jobs = [
    {'id': 'M1', 'volume': 100, 'priority': 2, 'print_time': 150},
    {'id': 'M2', 'volume': 150, 'priority': 1, 'print_time': 100},
    {'id': 'M3', 'volume': 120, 'priority': 3, 'print_time': 150}]


#тест перевищення обмежень об'єму
test3_jobs = [
    {'id': 'M1', 'volume': 250, 'priority': 1, 'print_time': 179},
    {'id': 'M2', 'volume': 200, 'priority': 1, 'print_time': 160},
    {'id': 'M3', 'volume': 180, 'priority': 2, 'print_time': 140}]


#обмеження принтера
constraints = {
    'max_volume': 300,
    'max_items': 2}


print('Тест 1 (однаковий пріоритет):')
result1 = optimize_printing(test1_jobs, constraints)

print(f'Порядок друку: {result1["print_order"]}')
print(f'Загальний час: {result1["total_time"]} хвилин')
print('--------------------------\n')


print('Тест 2 (різні пріоритети):')
result2 = optimize_printing(test2_jobs, constraints)

print(f'Порядок друку: {result2["print_order"]}')
print(f'Загальний час: {result2["total_time"]} хвилин')
print('--------------------------\n')


print('Тест 3 (перевищення обмежень):')
result3 = optimize_printing(test3_jobs, constraints)

print(f'Порядок друку: {result3["print_order"]}')
print(f'Загальний час: {result3["total_time"]} хвилин')
print('--------------------------\n')
```
